1 What is Python, and why is it popular
*   Python is a high-level, general-purpose programming language known for its simple syntax and readability. It is popular because it is easy to learn, supports multiple platforms, has extensive libraries, and is widely used in web development, automation, artificial intelligence, and data science.

2 What is an interpreter in Python
*   An interpreter in Python is a program that executes Python code by translating it into instructions the computer can understand.

3  What are pre-defined keywords in Python
*   Python keywords are reserved words that have predefined meanings and purposes in the Python language. They cannot be used as identifiers, such as variable names.

4  Can keywords be used as variable names
*   No, Python keywords cannot be used as variable names because they are reserved words with predefined meanings in the language.

5  What is mutability in Python
*   Mutability is the ability of an object to be modified after it has been created.  

6  Why are lists mutable, but tuples are immutable
*   Lists are mutable because Python allows their elements to be modified after creation. Tuples are immutable because their elements cannot be modified after creation. Lists are useful for data that changes frequently, whereas tuples are useful for fixed collections of data.

7 What is the difference between “==” and “is” operators in Python
*  The == operator checks whether two objects have the same value, whereas the is operator checks whether two variables refer to the same object in memory

8  What are logical operators in Python
*   Logical operators in Python are used to combine conditional statements and evaluate multiple conditions. Python has three logical operators: and, or, and not.

9 What is type casting in Python
*  Type casting in Python is the process of converting a value from one data type to another, such as converting a string to an integer or an integer to a float.

10  What is the difference between implicit and explicit type casting
*   Implicit type casting: Python automatically converts a value from one data type to another.
*   Explicit type casting: The programmer manually converts a value using functions such as int(), float(), and str().

11  What is the purpose of conditional statements in Python
*   Conditional statements in Python are used to make decisions in a program by executing different blocks of code based on whether a condition is True or False. Python supports if, elif, and else statements.

12 How does the elif statement work
*   The elif statement in Python stands for “else if.” It is used to check multiple conditions in sequence. If the if condition is false, Python checks the elif condition. If it is true, the corresponding block of code executes. If all conditions are false, the else block executes, if provided.

13  What is the difference between for and while loops
*   The for loop in Python is used to iterate over a sequence or iterable, such as a list, string, or range. The while loop is used to execute a block of code repeatedly as long as a specified condition is True.

14 Describe a scenario where a while loop is more suitable than a for loop.
*  A while loop is more suitable when the number of repetitions is not known in advance and depends on a condition. For example, when asking a user to enter a correct password, the loop continues until the correct password is entered. Since the number of attempts is unknown, a while loop is appropriate.

```python
#13 Write a Python program to find the sum of all even numbers between 1 and 50.
sum = 0
for i in range(1, 51):
  if i % 2 == 0:
    sum += i
print(sum)

```

```python
#14 Write a program to reverse a string using a while loop
string = input("Enter a string: ")
reverse = ""
i = len(string) - 1
while i >= 0:
  reverse += string[i]
  i -= 1
print(reverse)
```

```python
#15 Write a Python program to calculate the factorial of a number provided by the user using a while loop

n = int(input("Enter a number: "))
fact = 1

while n > 0:
    fact = fact * n
    n = n - 1

print(fact)
```
